```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
from sklearn.ensemble import RandomForestClassifier
import joblib
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier
```

```python
df = pd.read_csv("../data/processed_data/feature_train.csv")
```

```python
df.head()
```

```python
y = df["Churn"]
X = df.drop(columns = ["Churn"])
```

```python
y.unique()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2,stratify=y,random_state=42)
```

## Logistic Regression Models

```python
model1 = LogisticRegression()

model1.fit(X_train,y_train)

joblib.dump(model1,"../models/logistic.pkl")
```

## Random Forest Model

```python
model2 = RandomForestClassifier(n_estimators=100,random_state=42)

model2.fit(X_train,y_train)

joblib.dump(model2,"../models/Random_forest.pkl")
```

## XGBoost Model

```python
model3 = XGBClassifier(n_estimators=100,random_state = 42)

model3.fit(X_train,y_train)

joblib.dump(model3,"../models/xgboost.pkl")
```
