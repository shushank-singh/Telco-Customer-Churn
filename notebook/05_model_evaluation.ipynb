````python
import joblib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
````

````python
df = pd.read_csv("../data/processed_data/feature_train.csv")
````

````python
y = df["Churn"]
X = df.drop(columns = ["Churn"])
````

````python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2,stratify=y,random_state=42)
````

#### Load Saved Model From Models

````python
lr = joblib.load("../models/logistic.pkl")
rf = joblib.load("../models/Random_forest.pkl")
xgb = joblib.load("../models/xgboost.pkl")
````

````python
lr_pred = lr.predict(X_test)
rf_pred = rf.predict(X_test)
xgb_pred = xgb.predict(X_test)
````

#### Accuracy Scores For Each Models

````python
print("LR Accuracy:", accuracy_score(y_test, lr_pred))
print("RF Accuracy:", accuracy_score(y_test, rf_pred))
print("XGB Accuracy:", accuracy_score(y_test, xgb_pred))
````

#### Classification Report of Each Models

````python
print("LR Report", classification_report(y_test, lr_pred))
````

````python
print("RF Report", classification_report(y_test,rf_pred))
````

````python
print("XGB Report", classification_report(y_test,xgb_pred))
````

#### Confusion Matrix

````python
print("Logistic Regression")
print(confusion_matrix(y_test,lr_pred))
````
